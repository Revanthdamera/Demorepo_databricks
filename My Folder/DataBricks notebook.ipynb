```python
employees_table = spark.read.table("employees")
sales_customers = spark.read.table("samples.bakehouse.sales_customers")
```

```python
from pyspark.sql import functions as F
temp = employees_table.select("*").orderBy(F.col("id").asc())
temp2 = sales_customers.select("*").orderBy(F.col("customerID").asc())
temp.write.format("delta").mode("overwrite").saveAsTable("employees_delta")
temp2.write.format("delta").mode("overwrite").saveAsTable("sales_customers_delta")
```

```python
%sql
select * from employees_delta
```
